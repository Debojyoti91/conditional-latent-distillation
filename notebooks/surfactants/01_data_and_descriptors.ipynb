Check molecular identities before creating a new split. This demonstrates the safeguard without changing the reported cohort.

```python
from latent_materials.validation import grouped_development_split, assert_disjoint_molecules
example = ["CCO", "OCC", "CCC", "CCN", "NCC", "CCCC", "CCCl", "CCBr"]
train, validation = grouped_development_split(example, validation_fraction=0.3)
assert_disjoint_molecules([example[i] for i in train], [example[i] for i in validation])
train, validation
```
